```python

s12 = sp.Symbol('s12')
s23 = sp.Symbol('s23')
s34 = sp.Symbol('s34')
s45 = sp.Symbol('s45')
s51 = sp.Symbol('s51')


ns = {
    "s12": s12, "s23": s23,
    "dL": dL,
    "k1": kmom(1),
    "k2": kmom(2),
    "p1": pmom(1),
    "p2": pmom(2),
    "p3": pmom(3)}


k1, k2 = kmom(1), kmom(2)
p1, p2, p3, p4 = pmom(1), pmom(2), pmom(3), pmom(4)

kin = {dL(p1, p1): 0, dL(p2, p2): 0, dL(p3, p3): 0, dL(p1, p2): s12/2, dL(p2, p3): s23/2,  dL(p1, p3):(-s12-s23)/2}

kin = {dL(p1, p1): 0, dL(p2, p2): 0, dL(p3, p3): 0, dL(p1, p2): s12/2, dL(p2, p3): s23/2,  dL(p1, p3):(-s12-s23)/2}
dens = ([k1, k1-p1, k1-p1-p2, k1 - k2, k2-p1-p2, k2, k2-p1-p2-p3], [0, 0, 0, 0, 0, 0, 0], [k1, k2])
```
